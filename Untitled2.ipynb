```python
avg_Serum_Insulin=df_pyspark.agg(avg("Serum_Insulin")).collect()[0][0]
avg_bmi=df_pyspark.agg(avg("BMI")).collect()[0][0]
avg_glucose=df_pyspark.agg(avg("Glucose")).collect()[0][0]
avg_distolic=df_pyspark.agg(avg("Diastolic_BP")).collect()[0][0]
avg_skin=df_pyspark.agg(avg("Skin_Fold")).collect()[0][0]
```

```python
df_filled=df_pyspark.fillna({
    "Serum_Insulin":avg_Serum_Insulin,
    "BMI":avg_bmi,
    "Glucose":avg_glucose,
    "Diastolic_BP":avg_distolic,
    "Skin_Fold":avg_skin})
    
```

```python
df_filled=df_filled.withColumn("BMI_Category",
                               when(col("BMI") < 18.5,"underweight")
                               .when(col("BMI") < 25.0,"Normal")
                               .when(col("BMI") < 30,"overweight")
                               .otherwise("obese"))
```

```python
df_filled.show()
```

```python
df_filled=df_filled.withColumn("Age_Group",
                               when(col("Age")<18,"Teen")
                               .when(col("Age")<30,"young")
                               .when(col("Age")<40,"Adult")
                               .when(col("Age")<50,"middle")
                               .otherwise("senior"))
```

```python
df_filled.show()
```

```python
df_filled=df_filled.withColumn("Glucose_level",
                              when(col("Glucose") < 70,"low")
                              .when(col("Glucose") < 100,"Normal")
                              .when(col("Glucose") < 140,"Normal")
                              .otherwise("Diabetic range"))
```

```python
df_filled=df_filled.drop(col("Risk_Score"))
```

```python
df_filled.show()
```

```python
df_filled=df_filled.withColumn("Risk_Score",
                               round(
                                   (col("Glucose")*0.4)+
                                   (col("Glucose")*0.3)+
                                   (col("Age")*0.3),2))
                               
```

```python
df_filled.show()
```

```python
df_filled=df_filled.withColumn("Diagnosis",
                               when(col("Class")==1,"diabetic")
                               .otherwise("not diabetic"))
```

```python
df_filled.select(["Age","Age_Group","BMI","BMI_Category","Glucose","Glucose_level","Risk_Score","Diagnosis"]).show()
```

##AGGREGRATION

```python
df_filled.groupBy("Age_Group") \
.agg(
    count("*").alias("Total"),
    sum(when(col("Class")==1,1).otherwise(0)).alias("Diabetic"),
    sum(when(col("Class")==0,1).otherwise(0)).alias("non_diabetic")
).show()
```

```python
df_filled.groupBy("BMI_Category", "Class") \
    .count() \
    .orderBy("BMI_Category") \
    .show()




```

```python
df_filled.groupBy("Class") \
  .agg(
      count("*").alias("Total Patients"),
      round(avg("Glucose"), 2).alias("Avg Glucose"),
      round(avg("BMI"), 2).alias("Avg BMI"),
      round(avg("Age"), 2).alias("Avg Age"),
      round(avg("Serum_Insulin"), 2).alias("Avg Insulin"),
      round(avg("Diastolic_BP"), 2).alias("Avg BP"),
      max("Glucose").alias("Max Glucose"),
      min("Glucose").alias("Min Glucose")
  ).show()
```
