```python
df=df.withColumn("Diagnosis",
              when(col("Class")==1,"Diabetic")
              .otherwise("Non-Diabetic"))
```

```python
df.show()
```

```python
df=df.withColumn("BMI_Condition",
                 when(col("BMI")<18.5,"underweight")
                 .when(col("BMI")<25,"Normal")
                 .when(col("BMI")<30,"Overweight")
                 .otherwise("Obese"))
```

```python
df=df.withColumn("AGE_GROUP",
                 when(col("Age")<18,"teen")
                 .when(col("Age")<50,"Adult")
                 .otherwise("old"))
```

```python
df.show()
```

```python
df=df.withColumn("Glucose_level",
                 when(col("Glucose")<100,"normal")
                 .when(col("Glucose")<140,"pre diabetic")
                 .otherwise("high diabetic"))
```

```python
df.show()
```

```python
df.groupBy("Age_Group") \
.agg(
    count("*").alias("Total"),
    sum(when(col("Class")==1,1).otherwise(0)).alias("Diabetic"),
    sum(when(col("Class")==0,1).otherwise(0)).alias("non_diabetic")
).show()
```

```python
df.groupBy("Class") \
.agg(
    count("*").alias("total"),
    round(avg("Glucose"),2).alias("avg_glucose"),
    round(avg("BMI"),2).alias("avg_bmi")).show()
    
```

```python
df.groupBy("Diagnosis") \
.agg(
    max("BMI").alias("max_bmi"),
    min("BMI").alias("max_bmi")).show()
```

```python
df.groupBy("BMI_Condition") \
.agg(
    count("*").alias("total")).show()
```

```python
df.orderBy(["Age", "Glucose"], ascending=[True, False]).show()
```

```python
df.orderBy(["Age","Glucose"],ascending=[True,False]).show()
```

```python
df.orderBy("Age",ascending=True).show()
```

```python
df.groupBy("Class") \
.agg(
    round(avg("Glucose"),2).alias("avg_glucose")).orderBy("avg_glucose",ascending=False).show()
```

```python
df.groupBy("Diagnosis") \
.agg(
    sum(when(col("Glucose_level")=="normal",1).otherwise(0)).alias("total_normal"),
    sum(when(col("Glucose_level")=="pre diabetic",1).otherwise(0)).alias("total_pre_diabetic"),
    sum(when(col("Glucose_level")=="high diabetic",1).otherwise(0)).alias("total_high_diabetic")).show()

    
```

```python
df.groupBy("Class") \
.agg(
    count("*").alias("total_Count"),
    round(avg(col("BMI")),2).alias("avg_glucose"),
    max(col("Age")).alias("max_age")).show()
```
